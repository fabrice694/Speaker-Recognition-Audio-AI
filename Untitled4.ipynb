```python
import os
import librosa
import numpy as np
import joblib
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, accuracy_score

DATASET_PATH = '/content/drive/MyDrive/dataset feo'
CLASSES = ['Moi', 'Mon frère', 'Mon papa']

X = []
y = []

# Fonction amboarina tsara: Mamoaka MFCC 40 + Delta (Dynamics) + Normalization
def extract_robust_features(file_path, target_sr=22050):
    # Famakiana ilay feo sy fampitoviana ny Sample Rate
    audio, sr = librosa.load(file_path, sr=target_sr)

    # Esorina ny mangina (silence trimming) eo am-panombohana sy am-pamaranana
    audio, _ = librosa.effects.trim(audio, top_db=20)

    # Extract MFCC (40 coefficients)
    mfcc = librosa.feature.mfcc(y=audio, sr=target_sr, n_mfcc=40)
    mfcc_mean = np.mean(mfcc.T, axis=0)
    mfcc_std = np.std(mfcc.T, axis=0)

    # Delta (ny fiovan'ny feo amin'ny fotoana)
    delta_mfcc = librosa.feature.delta(mfcc)
    delta_mean = np.mean(delta_mfcc.T, axis=0)

    # Natambatra ho vecteur iray
    return np.hstack([mfcc_mean, mfcc_std, delta_mean])

print("⏳ Mamaky sy manatsara ny fitaovana amin'ny feo (Robust Feature Extraction)...")

for label in CLASSES:
    folder_path = os.path.join(DATASET_PATH, label)
    if not os.path.exists(folder_path):
        continue

    for file_name in os.listdir(folder_path):
        if file_name.lower().endswith(('.wav', '.mp3', '.m4a', '.ogg', '.webm')):
            file_path = os.path.join(folder_path, file_name)
            try:
                feat = extract_robust_features(file_path)
                X.append(feat)
                y.append(label)
                print(f"  + Voatrandraka tsara: {label} -> {file_name}")
            except Exception as e:
                print(f"  x Erreur: {file_name} -> {e}")

X = np.array(X)
y = np.array(y)

# Normalisation ho an'ny Machine Learning
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Model Random Forest matanjaka kokoa (n_estimators=200)
model = RandomForestClassifier(n_estimators=200, random_state=42)
model.fit(X_scaled, y)

# Tahirizina ilay Model sy ilay Scaler
joblib.dump(model, 'speaker_id_model.pkl')
joblib.dump(scaler, 'scaler.pkl')
print("\n✅ Voatahiry sy voafofana tsara ny AI vaovao!")
```

```python
import base64
import joblib
from IPython.display import display, HTML
from google.colab import output

# Charge-ina ny model sy ny scaler vaovao
model = joblib.load('speaker_id_model.pkl')
scaler = joblib.load('scaler.pkl')

def predict_speaker_robust(audio_path):
    feat = extract_robust_features(audio_path).reshape(1, -1)
    feat_scaled = scaler.transform(feat)

    pred = model.predict(feat_scaled)[0]
    probs = model.predict_proba(feat_scaled)[0]

    print(f"\n🎙️ Ny valiny: **{pred.upper()}** no miteny!")
    print("📊 Taux de confiance (Isan-jato):")
    for cls, prob in zip(model.classes_, probs):
        print(f"   - {cls}: {prob*100:.1f}%")

RECORD_JS = """
<script>
var recorder, stream, chunks = [];

async function startRecording() {
    chunks = [];
    stream = await navigator.mediaDevices.getUserMedia({ audio: true });
    recorder = new MediaRecorder(stream);
    recorder.ondataavailable = e => chunks.push(e.data);
    recorder.start();
    document.getElementById('status').innerText = '🎙️ Mitenena am-pitoniana amina mikrofona izao...';
    document.getElementById('btn-start').disabled = true;
    document.getElementById('btn-stop').disabled = false;
}

async function stopRecording() {
    recorder.stop();
    document.getElementById('status').innerText = '⏳ Fitsapana ny feo...';
    await new Promise(resolve => recorder.onstop = resolve);
    const blob = new Blob(chunks, { type: 'audio/wav' });
    const reader = new FileReader();
    reader.readAsDataURL(blob);
    reader.onloadend = () => {
        google.colab.kernel.invokeFunction('notebook.predict', [reader.result], {});
    };
}
</script>

<div style="padding: 10px; border: 1px solid #ccc; border-radius: 5px;">
  <button id="btn-start" onclick="startRecording()" style="padding: 10px 15px; background: #4CAF50; color: white; border: none; border-radius: 4px; cursor: pointer;">▶️️ Démarrer l'enregistrement</button>
  <button id="btn-stop" onclick="stopRecording()" disabled style="padding: 10px 15px; background: #f44336; color: white; border: none; border-radius: 4px; cursor: pointer; margin-left: 10px;">⏹️ Arrêter & Prédire</button>
  <p id="status" style="font-weight: bold; margin-top: 10px; color: #333;">Kitiho ny Démarrer rehefa vonona hiteny.</p>
</div>
"""

def process_audio(b64_data):
    audio_bytes = base64.b64decode(b64_data.split(',')[1])
    with open('test_mic_robust.wav', 'wb') as f:
        f.write(audio_bytes)

    print("\n--- VOKATRY NY FITILIANA VAOVAO ---")
    predict_speaker_robust('test_mic_robust.wav')

output.register_callback('notebook.predict', process_audio)
display(HTML(RECORD_JS))
```
